```python
import pandas as pd 
```

```python
df = pd.read_csv(r'orders.txt')
```

```python
dfr=pd.read_csv(r'returns.txt')
dfr.head(2)
```

```python
dff = pd.merge(left=df,right=dfr,how='left',left_on='order_id',right_on='order_id')
dff
```

```python
#dff[dff['Return_Reason'].isnull()].groupby('category')['sales'].sum()
rdf=dff[dff['Return_Reason'].isnull()]
rdf.groupby('category')['sales'].sum()
```

```python
#Q2
dff = pd.merge(left=df,right=dfr,how='inner',left_on='order_id',right_on='order_id')
dff
```
