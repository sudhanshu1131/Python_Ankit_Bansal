```python
import pandas as pd
```

```python
emp= {"name":["Ankit","Rahul","Priya"],"gender":["Male","Male","Female"], "email":["ankit@gmail.com","rahul@gmail.com","priya@gmail.com"] }
df1 = pd.DataFrame(emp)
df1
```

```python
emp= {"name":["Sudhanshu","vishal","Priyanshu"],"gender":["Male","Male","Female"], "email":["sudhanshu@gmail.com","vishal@gmail.com","priyanshu@gmail.com"] }
df2 = pd.DataFrame(emp)
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
#also use reset function
pd.concat([df1,df2]).reset_index()
```

```python

pd.concat([df1,df2], ignore_index=True,axis=1)
```

```python
emp= {"name":["Ankit","Rahul","Priya","vishal"],"gender":["Male","Male","Female","Male"], "email":["ankit@gmail.com","rahul@gmail.com","priya@gmail.com","ankit@gmail.com"] }
df = pd.DataFrame(emp)
df
```

```python
df.drop_duplicates()
```

```python
df.drop_duplicates(subset='email')
```

```python
df.drop_duplicates(subset='email',keep='first')
```

```python
df.drop_duplicates(subset='email',keep='last')
```

```python
df.drop_duplicates(subset=['email','gender'],keep='last')
#df.drop_duplicates(subset=['email','gender'],keep='last',inplace=true)
```

```python
df.drop_duplicates(subset=['email','gender'],keep=False)
```
